```python
customers['signup_date'] = pd.to_datetime(customers['signup_date'])
```

```python
subscriptions['subscription_start'] = pd.to_datetime(subscriptions['subscription_start'])
```

```python
subscriptions['subscription_end'] = pd.to_datetime(subscriptions['subscription_end'])
```

```python
payments['payment_date'] = pd.to_datetime(payments['payment_date'])
```

```python
usage['login_date'] = pd.to_datetime(usage['login_date'])
```

```python
tickets['created_date'] = pd.to_datetime(tickets['created_date'])
```

```python
churn['churn_date'] = pd.to_datetime(churn['churn_date'])
```

```python
customers['is_churned'] = customers['customer_id'].isin(churn['customer_id'])
```

```python
customers['age_group'] = pd.cut(
    customers['age'],
    bins=[18,25,35,45,55,65],
    labels=['18-25','26-35','36-45','46-55','56-65']
)
```

```python
subscriptions['revenue_segment'] = pd.cut(
    subscriptions['monthly_price'],
    bins=[-1,0,20,50,200],
    labels=['Free','Low Revenue','Medium Revenue','High Revenue']
)
```

```python
payments['revenue_month'] = payments['payment_date'].dt.to_period('M')
```

```python
monthly_revenue = payments.groupby(
    'revenue_month'
)['amount'].sum().reset_index()

print(monthly_revenue.head())
```

```python
customer_revenue = payments.groupby(
    'customer_id'
)['amount'].sum().reset_index()

customer_revenue.rename(
    columns={'amount':'total_customer_revenue'},
    inplace=True
)
```

```python
customers = customers.merge(
    customer_revenue,
    on= 'customer_id',
    how='left'
)
```

```python
customers.to_csv(
    "../Cleaned_Data/customers_cleaned.csv",
    index=False
)
```

##  REVENUE TREND ANALYSIS  ##


```python
monthly_revenue['revenue_month'] = monthly_revenue['revenue_month'].astype(str)

plt.figure(figsize=(12,6))

plt.plot(
    monthly_revenue['revenue_month'],
    monthly_revenue['amount']
)

plt.xticks(rotation=45)

plt.title("Monthly Revenue Trend")

plt.xlabel("Month")

plt.ylabel("Revenue")

plt.show()
```

```python
plan_revenue = subscriptions.groupby('plan_name')['monthly_price'].sum()

print(plan_revenue)
```

```python
colors = [
    '#38BDF8',  # Starter
    '#10B981',  # Pro
    '#8B5CF6',  # Enterprise
    '#F59E0B'   # Free
]

ax = plan_revenue.plot(
    kind='bar',
    figsize=(8,5),
    color=colors
)

plt.title("Revenue by Subscription Plan", fontsize=14, fontweight='bold')
plt.xlabel("Plan Name")
plt.ylabel("Revenue")

# Data Labels
for container in ax.containers:
    ax.bar_label(
        container,
        fmt='%.0f',
        padding=3
    )

# Clean look
plt.xticks(rotation=0)
plt.box(False)

plt.tight_layout()
plt.show()
```

```python
country_customers = customers['country'].value_counts()

print(country_customers)
```

```python
ax = country_customers.plot(
     kind='bar',
     figsize=(10,5)
)

plt.title("Customers by Country")

plt.ylabel("Customers")

#Add Data Labels
for container in ax.containers:
    ax.bar_label(
        container,
        padding=3
    )
    
plt.tight_layout()
plt.show()
```

```python
churn_summary = customers['is_churned'].value_counts()

print(churn_summary)
```

```python
ax = churn_summary.plot(
    kind='bar',
    figsize=(6,4),
    color=['#EF4444', '#10B981']
)

plt.title("Churn vs Active Customers", fontsize=14, fontweight='bold')
plt.ylabel("Customers")

for container in ax.containers:
    ax.bar_label(container, padding=3)

plt.box(False)
plt.tight_layout()
plt.show()
```

```python
churn_reason = churn['churn_reason'].value_counts()

print(churn_reason)
```

```python
ax = churn_reason.plot(
    kind='bar',
    figsize=(10,5),
    color=colors[:len(churn_reason)]
)

plt.title("Top Churn Reasons", fontsize=14, fontweight='bold')
plt.ylabel("Customers")

for container in ax.containers:
    ax.bar_label(container, padding=3)

plt.xticks(rotation=45, ha='right')
plt.box(False)
plt.tight_layout()

plt.show()
```

```python
age_analysis = customers.groupby(
    'age_group'
)['total_customer_revenue'].mean()

print(age_analysis)
```

```python
age_analysis.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title("Average Revenue by Age Group")

plt.ylabel("Average Revenue")

plt.show()
```
